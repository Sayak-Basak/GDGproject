```python
import os
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array

import os
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array

def load_data(data_dir, image_size=(224, 224)):
    images = []
    masks = []
    labels = []

    # Define paths for images and masks
    images_dir = os.path.join(data_dir, 'images')
    masks_dir = os.path.join(data_dir, 'masks')

    # Loop through each disease directory within images and masks
    for i, disease in enumerate(sorted(os.listdir(images_dir))):
        disease_images_dir = os.path.join(images_dir, disease)
        disease_masks_dir = os.path.join(masks_dir, disease)

        # Check if both disease directories exist
        if not os.path.isdir(disease_images_dir) or not os.path.isdir(disease_masks_dir):
            print(f"Warning: Missing directory for disease '{disease}' in images or masks. Skipping.")
            continue

        # Loop through each image in the disease-specific images directory
        for img_name in os.listdir(disease_images_dir):
            img_path = os.path.join(disease_images_dir, img_name)

            # Generate corresponding mask filename
            mask_name = img_name.replace('.jpg', '_mask.png')
            mask_path = os.path.join(disease_masks_dir, mask_name)

            # Check if the mask file exists
            if not os.path.exists(mask_path):
                print(f"Warning: Missing mask file for '{img_name}'''{mask_path}' in disease '{disease}'. Skipping.")
                continue

            # Load and preprocess the image and mask
            print(f"Loading '{img_name}'")
            img = load_img(img_path, target_size=image_size)
            img = img_to_array(img) / 255.0  # Normalize image

            mask = load_img(mask_path, target_size=image_size, color_mode='grayscale')
            mask = img_to_array(mask) / 255.0  # Normalize mask

            images.append(img)
            masks.append(mask)
            labels.append(i)  # Append the disease label

    return np.array(images), np.array(masks), np.array(labels)

# Define paths for training, validation, and testing sets
train_dir = r'dataset\train'
val_dir = r'dataset\val'
test_dir = r'dataset\test'

# Load data for each set
X_train, mask_train, y_train = load_data(train_dir)
X_val, mask_val, y_val = load_data(val_dir)
X_test, mask_test, y_test = load_data(test_dir)

y_train = np.expand_dims(y_train, axis=-1)  # Add channel dimension for labels
y_val = np.expand_dims(y_val, axis=-1)
y_test = np.expand_dims(y_test, axis=-1)
```
